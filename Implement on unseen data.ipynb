```python
import pandas as pd
import joblib

# Load saved pipeline
model = joblib.load("models/student_model.pkl")

# Load new unseen data
new_data = pd.read_csv(r"C:\Users\keert\Downloads\new_students.csv")

# Select the 19 features expected by the model
X_new = new_data[list(model.feature_names_in_)]

# Predict directly without manual encoding
predictions = model.predict(X_new)

# Add predictions to the data
new_data["Predicted_Exam_Score"] = predictions.round(2)

# Display results
display(new_data[[
    "Hours_Studied",
    "Attendance",
    "Previous_Scores",
    "Predicted_Exam_Score"
]])

# Save results
new_data.to_csv("student_predictions.csv", index=False)
```
